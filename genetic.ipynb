```python
import random
import numpy as np

random.seed(42)
sol_per_pop = 8
num_genes = 4

population = np.random.uniform(low=-1, high=1, size=(sol_per_pop, num_genes))

for i, chromosome in enumerate(population):
    print(f"\nChromosome {i+1}")
    for j, gene in enumerate(chromosome):
        print(f"  Gene {j+1} = {gene:.2f}")

min_val = np.min(population)
max_val = np.max(population)
avg_val = np.mean(population)

print(f"\nMin gene value : {min_val:.3f}")
print(f"Max gene value : {max_val:.3f}")
print(f"Average value  : {avg_val:.3f}")

fitness = [] 
for i in range (sol_per_pop):
    total_fitness = 0
    print("\n Chromosomes" , i+1)

    for j in range( num_genes):
        gene = population[i][j]
        gene_fitness = gene**2

        print("gene" , j+1  , "= %.3f = %.3f " % (gene , gene_fitness ))
        total_fitness += gene_fitness

    fitness.append(total_fitness)
    print("total fitness = " , round(total_fitness , 3))
```
